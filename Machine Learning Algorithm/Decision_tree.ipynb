```python


import numpy as np
import matplotlib.pyplot as plt

def make_data() :
  X = np.array( [
    [165, 19], [175, 32], [136, 35], [174, 65], [141, 28],
    [176, 15], [131, 32], [166, 6], [128, 32], [179, 10],
    [136, 34], [186, 2], [126, 25], [176, 28], [112, 38],
    [169, 9], [171, 36], [116, 25], [196, 25], [196, 38],
    [126, 40], [197, 20], [150, 25], [140, 32], [136, 35] ] )
  Y = np.array([1, 2, 2, 1, 2, 1, 2, 1, 2, 1, 2, 1, 2, 2, 2, 1, 2, 2, 1, 2, 2, 1, 1, 2, 2])
  return X, Y


X, Y = make_data()
for index in range(len(X)) :
  coordinate = X[index]
  xx = coordinate[0]
  yy = coordinate[1]
  if Y[index] == 1 :
    plt.scatter(xx, yy, c='blue')
  else :
    plt.scatter(xx, yy, c='red')
plt.show()
```

```python
import numpy as np
from sklearn import tree
import matplotlib.pyplot as plt
from mlxtend.plotting import plot_decision_regions
X, Y = make_data()
DT = tree.DecisionTreeClassifier(criterion='entropy', max_depth=4)
DT = DT.fit(X, Y)
plot_decision_regions(X, Y, clf = DT)
plt.show()
```
